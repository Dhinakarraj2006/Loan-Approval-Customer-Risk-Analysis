🔹 Section 2 — Credit Score Analysis

Q3. How does the loan approval rate vary across Low, Medium, and High credit-score tiers?

SELECT
    CASE
        WHEN Credit_Score < 600 THEN 'Low (<600)'
        WHEN Credit_Score BETWEEN 600 AND 750 THEN 'Medium (600-750)'

Q4. What is the approval and rejection rate for applicants with a credit score below 600?

        CASE
            WHEN Loan_Approval_Status = 1 THEN 1

            ELSE 0

Q8. What percentage of the total loan portfolio belongs to this high-approval segment?

SQL Query

WITH HighApprovalSegment AS (
    SELECT
        COUNT(\*) AS high_approval_count
    FROM loans
    WHERE Credit_Score >= 600
      AND Age <= 50
),
TotalPortfolio AS (
    SELECT
        COUNT(*) AS total_count
    FROM loans
)
SELECT
    h.high_approval_count,
    t.total_count,
    ROUND(
        h.high_approval_count * 100.0 / t.total_count,
        2
    ) AS portfolio_percentage
FROM HighApprovalSegment h
CROSS JOIN TotalPortfolio t;

    SELECT
        Applicant_ID,
        Employment_Status,
        Annual_Income,
        Loan_Amount_Requested,
        Loan_Approval_Status,
        DENSE_RANK() OVER (
            PARTITION BY Employment_Status
            ORDER BY Annual_Income DESC
        ) AS income_rank
    FROM loans
    WHERE Loan_Approval_Status = 1
)
SELECT
    Applicant_ID,
    Employment_Status,
    Annual_Income,
    Loan_Amount_Requested,
    income_rank
FROM RankedApplicants
WHERE income_rank <= 3
