```python
loan[loan['Age'] <= 50]['Loan_Approval_Status'].value_counts(normalize=True) * 100
```

```python
loan[(loan['Credit_Score'] >= 600) & (loan['Age'] <= 50)]['Loan_Approval_Status'].value_counts(normalize=True) * 100
```

```python
features = ['Gender', 'Marital_Status', 'Education', 'Employment_Status', 'Residential_Status', 'City_Town', 'Loan_Purpose', 'Loan_Type']

for col in features:
    display(loan.groupby(col)['Loan_Approval_Status'].value_counts(normalize=True) * 100)
```

```python
loan.groupby('Loan_Approval_Status')['Loan_Amount_Requested'].mean()
```

```python
loan[(loan['Credit_Score'] >= 600) & (loan['Age'] <= 50)]['Loan_Approval_Status'].value_counts(normalize=True) * 100
```

```python
loan.groupby('Loan_Approval_Status')[['Annual_Income', 'Outstanding_Debt', 'Default_Risk']].mean()
```
